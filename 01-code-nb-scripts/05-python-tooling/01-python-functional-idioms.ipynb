### Python Functional IdiomsPassing a small function as a value: lambda expressions, then `operator.itemgetter` as the faster, more idiomatic alternative to a lambda for the common "sort/access by key" case.

#### Lambda functions

```python
lambda x : x
# Lambda functions are functions

# let's comapre a lambda function with a standard function
# lambda func are anyonymous. They don't have a name
# do not have return keyword. Whatever follows the colon is the expression

# You can have only single that evaluated to a value.
# This is value returned by lambda function

def do_boring_stuff(x):
    return 42 * x

lambda x : 42 * x
```

```python
print(do_boring_stuff(10))
print((lambda x : 42 * x)(10)) # enclose the expression in parentheses
```

```python
# To name a lambda function -
name_lambda_func = lambda x : 42 * x
print(name_lambda_func(10))
```

```python
print((lambda x, y: x * y)(10, 2))
print((lambda x, y, z: x * y + z)(10, 2, 5))
```

```python
scores = {
    "Stephen": 8,
    "James": 4,
    "Sarah": 9,
    "Kate": 6
}

def check_pass(item):
    return scores[item] >= 5

outcomes = map(check_pass, scores)
# each key in dict score is passed to check_pass

print(list(outcomes))

# Why are the dictionary keys passed to the function and not the key-value pairs?
### -> Iteration over dictionaries in Python uses the keys. This is also the case if you use a dictionary in a for loop,

outcomes = map(lambda item: scores[item] >= 5, scores)
print(list(outcomes))

# More Pythonic & Preferrable
outcomes = [scores[item] >=5 for item in scores]
print(outcomes)
```

#### `operator.itemgetter`: sort-key benchmarks vs. lambda vs. a named function

```python
team_members = [
    ("Kate", 8),
    ("John", 5),
    ("Alice, 10"),
    ("Bob, 7"),
    ("Eve, 6"),
    ("Charlie, 9"),
]

team_members[3]

# You can get the fouthr item using operator.itemgetter()
# Returns a callable, an then you can pass the list to that callable

from operator import itemgetter
get_fourth = itemgetter(3) # Returns a callable
get_fourth(team_members) # will always fetch fourth item

# Or whatever subscriptable if you can use square bracket notations to fetch

```

```python
class Team:
    def __init__(self, team_iterable):
        self.team = list(team_iterable)
    
    def __getitem__(self, key_or_index):
        print("Hey look ! __getitem__ was called behind the scenes")
        return self.team[key_or_index]
    
team = Team(team_members)
team[3]
```

```python
team_members = [
    ("Kate", 8),
    ("John", 5),
    ("Alice", 10),
    ("Bob", 7),
    ("Eve",6)
 ]

# Key parameter that you can use to set your own sorting rules
# Argument you pass to this key parameter should be function (Callable)
sorted(team_members, key=lambda x : x[1])

from operator import itemgetter
get_score = itemgetter(1)
sorted(team_members, key=get_score)
```

```python
import timeit
timeit.timeit(
    "sorted(team_members, key=get_score)",
    number=1_000_000,
    globals=globals()
)
```

```python

timeit.timeit(
    "sorted(team_members, key=itemgetter(1))",
    number=1_000_000,
    globals=globals()
)
```

```python
timeit.timeit(
    "sorted(team_members, key=lambda x: x[1])",
    number=1_000_000,
    globals=globals()
)
```

```python
members = [
    {"name": "Josh", "membership_tier": 1, "year_joined": 2018},
     {"name": "Sally", "membership_tier": 2, "year_joined": 2020},
     {"name": "Mike", "membership_tier": 1, "year_joined": 2019},
     {"name": "Anna", "membership_tier": 2, "year_joined": 2017},
     {"name": "Tom", "membership_tier": 1, "year_joined": 2018},
     {"name": "Lucy", "membership_tier": 2, "year_joined": 2020},
 ]

from operator import itemgetter
sort_order = itemgetter("membership_tier", "year_joined", "name")
# can accept multiple arguments.

sort_order(members[0])
# returns a tuple with values in order of keys

sorted(members, key=sort_order)
```
