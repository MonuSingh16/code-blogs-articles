## Cyclical Features

1. Cyclical features are periodic, unlike numerical (age, amount) or categorical (t-shirt size) ones: hour of day (0..23, then repeats), day of week, month of year.
2. A plain linear encoding fails: "23" is numerically far from "0" even though they're adjacent on the clock, and the distance property isn't preserved anywhere else on the cycle either.
3. Fix: map onto a circle with sine and cosine. For period $T$: normalize $x_{norm}=x/T$, angle $\theta=2\pi x_{norm}$, encode as $(\sin\theta,\cos\theta)$.
4. Why both, not one: $\sin(0°)=\sin(180°)=0$ — sine alone can't distinguish opposite points, same problem for cosine alone. 90° out of phase (orthogonal), so together they give a unique point for every angle on the circle.
5. Benefits: preserves true cyclic distance, no discontinuity at the wrap-around boundary, smooth for a model to learn, avoids the model treating high/low values as extremes (e.g. wrongly learning "late hours are bad" when it's just proximity to midnight).
6. Used for: time-series forecasting, recommendation systems (activity by time), fraud detection (unusual-hour transactions), geospatial (wind direction), retail seasonality, IoT sensor cycles — standard practice for any time feature in an ML pipeline, usually combined with other time features like holidays.


```python
import numpy as np
import pandas as pd

# Create sample data
hours = [0, 6, 12, 18, 23, 24]  # 24 wraps to 0
df = pd.DataFrame({'hour': hours})

# Encode cyclically
df['sin_hour'] = np.sin(df['hour'] * 2 * np.pi / 24)
df['cos_hour'] = np.cos(df['hour'] * 2 * np.pi / 24)

print(df)
```

```python
import numpy as np
import matplotlib.pyplot as plt

hours = np.arange(24)
sin_hours = np.sin(hours * 2 * np.pi / 24)
cos_hours = np.cos(hours * 2 * np.pi / 24)

plt.figure(figsize=(6,6))
plt.scatter(sin_hours, cos_hours, c=hours, cmap='hsv')
plt.xlabel('sin(hour)')
plt.ylabel('cos(hour)')
plt.title('Cyclical Encoding of Hours')
plt.axis('equal')
for i, h in enumerate(hours):
    plt.annotate(str(h), (sin_hours[i], cos_hours[i]))
plt.show()
```
