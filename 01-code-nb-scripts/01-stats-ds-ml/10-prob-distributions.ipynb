#### Variance

Q1. You play a game with a friend, You win 1 Dollar if it is heads and lose 1 dollar if it is tail. What is fair amount of money to pay to play this game ?

- This is a fair coin-flip bet, so the expected value (EV) is: EV = (0.5 × +$1) + (0.5 × −$1) = $0.50 − $0.50 = $0

- Since the expected value is zero, the "fair" price to play is $0 — no one should have to pay anything to enter, and neither side has an edge. This is the essence of a martingale (fair game) in probability theory.

##### Measuring Spread - Variance Motivation

Expected Squared Deviation = E[(X- E[X])^2], This is variance. There are 4 steps to this
1. Find X's Mean
2. Find the deviation from that mean for every value of X
3. Square those deviations
4. Average those squared deviations

Var(X) = E[(X- E[X])^2] = E[X^2] - E[X]^2

We can arrive to this by 

$$Var(X) = \mathbb{E}[(X - \mathbb{E}[X])^2] = \mathbb{E}[X^2 - 2\mathbb{E}[X]X + \mathbb{E}[X]^2]$$
$$= \mathbb{E}[X^2] - \mathbb{E}[2\mathbb{E}[X]X] + \mathbb{E}[\mathbb{E}[X]^2]$$
$$= \mathbb{E}[X^2] - 2\mathbb{E}[X]\,\mathbb{E}[X] + \mathbb{E}[X]^2$$
$$= \mathbb{E}[X^2] - 2\mathbb{E}[X]^2 + \mathbb{E}[X]^2$$
$$= \mathbb{E}[X^2] - \mathbb{E}[X]^2$$

*Rules used*:
- $\mathbb{E}[X+Y] = \mathbb{E}[X] + \mathbb{E}[Y]$
- $\mathbb{E}[\text{constant} \cdot X] = \text{constant} \cdot \mathbb{E}[X]$
- $\mathbb{E}[X]$ is a constant
- $\mathbb{E}[\text{constant}] = \text{constant}$

Another One. 
- $Var(ax + b) = a^2Var(x)$

#### Practice Q&A — Expected Value & Variance

**Q1.** You roll a fair 6-sided die. You win $X$ dollars where $X$ is the number shown. What is $E[X]$?

- $E[X] = \frac{1}{6}(1+2+3+4+5+6) = \frac{21}{6} = 3.5$
- Fair price to play = **$3.50**

---

**Q2.** Same die game as Q2. What is $Var(X)$?

- $E[X^2] = \frac{1}{6}(1^2+2^2+3^2+4^2+5^2+6^2) = \frac{91}{6} \approx 15.17$
- $Var(X) = E[X^2] - E[X]^2 = 15.17 - (3.5)^2 = 15.17 - 12.25 = 2.92$

---

**Q3.** If you win $2 for heads and lose $1 for tails (biased payout, fair coin), what is the fair price to play?

- $E[X] = 0.5(2) + 0.5(-1) = 1 - 0.5 = 0.50$
- Fair price = **$0.50** (not $0 this time — payout is asymmetric even though the coin is fair)
- 💡 Key insight: "fair coin" ≠ "fair game." Fairness of the *game* depends on the **payout structure**, not just the coin.

---

**Q4.** Why is $Var(X) \geq 0$ always?

- $Var(X) = E[(X-E[X])^2]$ — it's an expectation of a **squared** term, and squares are never negative.
- So the average of non-negative numbers must itself be non-negative.

---

**Q5.** If $X$ is a constant (say $X = 5$ always), what is $Var(X)$?

- $Var(X) = 0$
- Makes sense: **no spread → no variance**. Variance measures *uncertainty*; a constant has none.

---

**Q6.** Given $Var(X) = a^2 Var(X)$... wait — prove $Var(aX+b) = a^2 Var(X)$.

$$Var(aX+b) = E[(aX+b)^2] - (E[aX+b])^2$$
$$= E[a^2X^2 + 2abX + b^2] - (aE[X]+b)^2$$
$$= a^2E[X^2] + 2abE[X] + b^2 - \left(a^2E[X]^2 + 2abE[X] + b^2\right)$$
$$= a^2E[X^2] - a^2E[X]^2 = a^2\big(E[X^2]-E[X]^2\big) = a^2Var(X)$$

- 💡 **Intuition:** adding $b$ (a shift) doesn't change spread → $b$ vanishes. Scaling by $a$ stretches spread by $a$, but since variance is in *squared* units, it scales by $a^2$.
- This is why **standard deviation** ($\sigma = \sqrt{Var(X)}$) scales by $|a|$, not $a^2$ — it's in the original units.

---

**Q7.** Two independent games: Game A has $Var(X)=4$, Game B has $Var(Y)=9$. What is $Var(X+Y)$?

- For **independent** $X, Y$: $Var(X+Y) = Var(X) + Var(Y) = 4 + 9 = 13$
- ⚠️ This only holds under independence. In general: $Var(X+Y) = Var(X)+Var(Y)+2\,Cov(X,Y)$

---

**Q8.** Quick check — is a $0 fair price always the "best" game to play?

- No. EV=0 means no long-run edge, but **variance** still matters for risk. A game with EV=$0$ and $Var=100$ feels very different from one with EV=$0$ and $Var=1$, even though both are "fair" by the EV definition.
- This is why real decision-making often uses **risk-adjusted** measures, not EV alone (see earlier note on utility theory / St. Petersburg Paradox).
