## Variance

1. Fair-price intuition: a coin-flip bet (+$1 heads, -$1 tails) has $EV=(0.5\times1)+(0.5\times-1)=0$, expected value zero means $0 is the fair price to play, neither side has an edge.
2. Variance is expected squared deviation: $Var(X)=E[(X-E[X])^2]$. Four steps: find the mean, find each value's deviation from it, square the deviations, average them.
3. Shortcut form, derived: $E[(X-E[X])^2]=E[X^2-2E[X]X+E[X]^2]=E[X^2]-2E[X]^2+E[X]^2=E[X^2]-E[X]^2$. Rules used: $E[X+Y]=E[X]+E[Y]$; $E[cX]=cE[X]$; $E[X]$ is a constant; $E[c]=c$.
4. Scaling rule: $Var(aX+b)=a^2Var(X)$, shifting by $b$ doesn't change spread, scaling by $a$ scales variance by $a^2$ (not $a$), since deviations get squared.
