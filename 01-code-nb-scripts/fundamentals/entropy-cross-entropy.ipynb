## Entropy and Cross-Entropy

1. Information = surprise. Rain in a rainforest: ~0 information (expected). Rain in the Sahara: high information (unexpected).
2. Entropy $H$ = average surprise/uncertainty in a system. Always-heads coin: 0 entropy, never surprised. Fair coin: maximum entropy, maximally uncertain before the flip.
3. Cross-entropy $H(p,q)$ = extra surprise from using a wrong belief $q$ to predict the real distribution $p$. $\text{Cross-Entropy}=\text{Entropy}+\text{KL Divergence}$. If $q=p$, cross-entropy = entropy (no extra surprise). If $q\neq p$, cross-entropy exceeds entropy.
4. Cross-entropy in bits: perfect prediction uses the minimum bits (entropy); wrong prediction uses extra bits, and that extra is the error.
5. Training = pushing predicted $q$ toward true $p$; cross-entropy as a loss function is a single number for "how much extra surprise is my model creating."
6. Hard labels (100% yes/no) have entropy 0, no uncertainty in the ground truth. So $\text{Cross-Entropy}=0+\text{Error}=\text{Error}$: predicting 0.7 for a true label of 1.0, the cross-entropy value *is* exactly the error from that 0.3 gap.
