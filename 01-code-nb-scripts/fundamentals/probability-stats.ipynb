## Probability & Stats

Organized in the order it's most naturally learned: combinatorics → distributions → estimation paradigms → hypothesis testing.

### Combinatorics & Conditional Probability

**Conditional Probability via Direct Counting**

1. Setup: 200 people — 30 soccer-only, 100 basketball-only, 70 both. Want P(basketball | soccer).
2. Trigger phrase: "probability of X given Y" → conditional probability.
3. P(soccer) = only-soccer + both = 30 + 70 = 100.
4. P(basketball ∩ soccer) = both = 70.
5. P(basketball | soccer) = P(both) / P(soccer) = 70/100 = 0.7.
6. "Given they like soccer" shrinks the sample space to the 100 soccer-likers — direct counting, one step before needing the full Bayes formula (needed only when given reverse conditionals, not raw counts like this).

**Binomial Distribution — Fixed Trials, Exact Count of Successes**

1. Setup: roll a die 20 times, want P(exactly 7 show a 4).
2. Trigger phrases: "fixed number of trials", "exactly k successes", "probability X appears N times".
3. Why Binomial: fixed n=20 trials, independent, fixed p=1/6, counting successes k=7 — the exact setup Binomial is built for.
4. Formula: P(X=k) = C(n,k)·p^k·(1-p)^(n-k) → P(X=7) = C(20,7)·(1/6)^7·(5/6)^13.
5. C(n,k) = ways to choose which trials succeed; p^k = chosen trials succeed; (1-p)^(n-k) = remaining trials fail.
6. Trap: using (1/6)^20 instead of (1/6)^7, or dropping C(n,k) entirely.

**Bayes' Theorem — Base Rate Neglect**

1. Setup: a disease affects 1% of the population; the test is 95% sensitive, 90% specific. Given a positive test, what's P(sick)?
2. Given: P(sick)=0.01, P(not sick)=0.99, P(pos\|sick)=0.95, P(neg\|not sick)=0.90 → P(pos\|not sick)=0.10.
3. Law of Total Probability: P(pos) = (0.95)(0.01) + (0.10)(0.99) = 0.1085.
4. Bayes: P(sick\|pos) = 0.0095 / 0.1085 ≈ 8.76%.
5. Key idea: even a 95%-accurate test yields a low true-positive rate when the base rate is low — most positives come from the much larger healthy population's false positives, not actual sick people.
6. Always draw the numbers into a table/tree first to avoid confusing P(sick\|pos) with P(pos\|sick).

**Complement Rule — 'Not All' Events**

1. Setup: throw 10 fair coins, want P(not all heads).
2. P(all heads) = (1/2)^10 = 1/1024.
3. Complement: P(not all heads) = 1 − 1/1024 ≈ 99.90%.
4. Far easier via the complement than summing every other outcome (0–9 heads) — avoids binomial sums entirely, since there's exactly one way to get "all heads" out of 2^10 total outcomes.

**Complement Rule — Threshold at the Minimum Possible Value**

1. Setup: roll 10 dice, sum the results, want P(sum > 10).
2. Don't compute the full sum distribution — check the boundary first: the minimum possible sum of 10 dice = 10 (all show 1).
3. P(sum=10) = P(all ten show 1) = (1/6)^10, the only way to hit the minimum exactly.
4. P(sum>10) = 1 − (1/6)^10 ≈ 99.99998%.
5. Recognition signal: whenever a threshold coincides with the minimum possible value of a sum, check if the complement collapses to one trivial outcome — same trick as the coin question above.

**Expected Value vs. Variance — Comparing Two Bets**

1. Setup: Game A: 1/3 chance win \$2, 2/3 chance lose \$1. Game B: four outcomes mixing smaller and larger wins/losses.
2. Formula: E[X] = Σxᵢ·pᵢ.
3. Game A: E[A] = (2)(1/3) + (−1)(2/3) = \$0 — a fair game.
4. Game B: E[B] = 0.25 − 0.125 + 0.625 − 0.25 = \$0.50.
5. Game B has the higher EV despite more losing outcomes by count — the rare \$5 win carries enough weight to outpull the other three outcomes combined.
6. EV alone ignores risk: Game B's outcomes swing much wider (\$5 win, \$2 loss) than Game A's (\$2/\$1) — a risk-averse player might still prefer Game A despite the lower EV. Worth flagging this trade-off, not just the EV number.
7. Kurtosis measures tail thickness — Game B's larger kurtosis comes from its extreme, low-probability \$5-win/\$2-loss outcomes.

### Distributions & Descriptive Statistics

**Reading a Discrete Probability Table**

1. Setup: passenger counts 0–6 with given probabilities; want P(X ≤ 3).
2. P(X ≤ 3) = p(0) + p(1) + p(2) + p(3) = 0.10 + 0.25 + 0.25 + 0.15 = 0.75.
3. For a discrete table, "≤ k" just means summing every row up to and including that value — no CDF formula or distribution assumption needed, since the full probability mass function is given directly.

**Central Limit Theorem vs. Law of Large Numbers**

1. CLT formula: sample mean ~ Normal(μ, σ²/n), so SE = σ/√n.
2. CLT intuition: as n grows, the sample mean's distribution becomes Normal-shaped regardless of the population's shape — why Normal-based inference (z/t-tests, CIs) works even on skewed data.
3. CLT example: die rolls (uniform, not Normal) — the average of 30 rolls, repeated many times, forms a Normal curve centered at 3.5, spread shrinking as n grows.
4. LLN formula: x̄ₙ → μ as n → ∞ (convergence in probability).
5. LLN intuition: more data makes the sample mean a more accurate estimate of the true mean — about accuracy, not shape. Example: a coin might show 70% heads after 10 flips, converging close to 50% after 10,000.
6. Key distinction: LLN = where you're heading (accuracy). CLT = the shape/spread of the estimate around that point (Normal, shrinking SE) — same n→∞ setting, different questions.

**Law of Large Numbers — Worked Example & Real-World Uses**

1. Formula: x̄ₙ → μ as n → ∞ (the sample mean converges in probability to the true population mean).
2. Worked example (fair coin, true μ=0.5): n=10 → 7 heads → sample mean 0.70 (far from 0.5); n=100 → 53 heads → 0.53 (closer); n=10,000 → 5,012 heads → 0.5012 (very close) — a trend you can trace numerically, not a formula you "solve."
3. Real-world use: A/B testing (larger n → more trustworthy conversion-rate estimate); insurance (individual claims unpredictable, but the average claim cost stabilizes across a large policyholder pool); Monte Carlo simulation (the average of many random trials converges to the true expected value).

**Right Skew — Why Mean Exceeds Median**

1. Relationship: right-skewed (positive skew) → mean > median > mode.
2. Intuition: the mean is pulled toward the long tail because it factors in every value's magnitude (including extreme outliers), while the median only cares about the middle position/rank.
3. Worked example: salaries \$40k, \$45k, \$50k, \$55k, \$500k (one outlier). Median = \$50k (unaffected by how extreme \$500k is); mean = \$138k (heavily pulled up). Mean >> median is the classic right-skew signature.
4. Real-world use: income/salary/house-price data is usually reported as median, not mean — median better represents the "typical" value when a distribution has a long right tail of high outliers.

**Skewness — Measuring Asymmetry**

1. Formula: Skewness = E[(X-μ)³] / σ³ (third standardized moment).
2. Intuition: measures asymmetry around the mean. Positive skew = long right tail (mean pulled above median); negative skew = long left tail (mean pulled below median); zero skew = symmetric (like Normal).
3. Continuing the salary example (mean \$138k >> median \$50k → positive skew): the \$500k outlier's cubed deviation dominates the sum, producing a large positive skewness (≈2.0, strongly right-skewed).
4. Rule of thumb: skew ≈ 0 roughly symmetric; skew > 0 right-skewed; skew < 0 left-skewed; \|skew\| > 1 considered highly skewed.
5. Real-world use: check skewness before choosing mean vs. median as a summary statistic, or before applying tests (like a t-test) that assume roughly symmetric data — highly skewed data often needs a log transform or a non-parametric test instead.

### Estimation & Inference Paradigms

**Probability vs. Likelihood**

1. Probability = P(data \| parameters) — parameters fixed, data varies. Likelihood = L(parameters \| data) — data fixed, parameters vary (same mathematical function, just which side you hold fixed flips).
2. Probability asks "given a known model, how likely is this outcome?" Likelihood asks "given this observed outcome, how plausible is each possible model?" — probability predicts future data from a known model; likelihood infers the model from data already in hand.
3. Worked example: a coin with P(heads)=p, flipped 10 times, 7 heads observed. Probability question: "if p=0.5, what's P(exactly 7 heads)?" → plug into Binomial, ≈0.117. Likelihood question: "given 7 heads observed, how plausible is p=0.5 vs. 0.7 vs. 0.9?" → compare the Binomial formula at k=7 across candidate p's; here L(p=0.7) > L(p=0.5), suggesting p=0.7 fits the data better.
4. This distinction underlies Maximum Likelihood Estimation — "fitting a model" (e.g. logistic regression's coefficients) means finding the parameters that maximize the likelihood of the observed data.

**Maximum Likelihood Estimation**

1. MLE picks whichever parameter setting makes the data actually observed the most probable one to have occurred — reverse-engineering the most plausible explanation from the evidence.
2. Formula: θ̂_MLE = argmax_θ P(data \| θ) = argmax_θ Σᵢ log P(xᵢ \| θ).
3. Procedure: pick a model family (Normal, Poisson, ...) believed to have generated the data; write the data's probability as a function of the unknown parameter(s); find the value that maximizes it (derivative = 0, solve).
4. Examples: logistic regression coefficients are fit via MLE (maximize the likelihood of the observed 0/1 labels); fitting any distribution's parameters to data (e.g. λ for a Poisson from observed event counts) uses the same argmax-log-likelihood procedure.
5. Limitation worth knowing: MLE can overfit with small samples (1 flip, 1 head → p̂=1.0, clearly overconfident) — exactly where MAP estimation (Bayesian, incorporates a prior) is preferred, since it regularizes toward plausible values with limited data.

**Frequentist vs. Bayesian Inference**

1. Frequentist: parameters are fixed, unknown constants; probability = long-run frequency over repeated experiments.
2. Bayesian: parameters are random variables with their own probability distributions; probability = degree of belief, updated as evidence arrives.
3. Bayesian core formula: Posterior ∝ Likelihood × Prior, i.e. P(θ \| data) ∝ P(data \| θ) · P(θ). Frequentist has no equivalent — it never treats θ as something with a probability distribution.
4. Frequentist question: "if I repeated this experiment infinitely, what fraction of the time would this procedure be right?" — inference is about the procedure's long-run behavior (same idea behind 95% CI coverage).
5. Bayesian question: "given what I believed before, and the data I just saw, what do I believe now?" — inference directly updates a belief distribution over the parameter itself.

### Hypothesis Testing & Experiment Design

**What a P-Value Actually Means**

1. Definition: p-value = P(observing data this extreme or more \| H₀ is true) — a conditional probability computed assuming the null hypothesis is true.
2. It tells you how surprising the data would be if there were truly no effect, not the probability that there's no effect.
3. Why it is NOT P(H₀ true): the p-value is P(data \| H₀ true); what's often (wrongly) wanted is P(H₀ true \| data) — not the same thing, confusing the two is the "prosecutor's fallacy" applied to statistics. Getting P(H₀ true \| data) requires Bayes' Theorem plus a prior P(H₀), which a p-value never incorporates.
4. Correct interpretation: p = 0.03 means "if there truly were no effect, we'd see data this extreme (or more) only 3% of the time by chance" — not "3% chance the null is true."
5. Common misconception to preempt: a small p-value doesn't measure effect size either — a tiny, meaningless effect can still produce p<0.05 with a large enough sample.

**Welch's Two-Sample T-Test**

1. H0: μ1=μ2 (no difference). H1: μ1≠μ2. Two independent groups, not paired.
2. Welch's vs. Student's: Student's assumes equal variance between groups; Welch's doesn't — the safer default, since real groups rarely have equal variance.
3. t-stat: t = (m1-m2) / SE, SE = sqrt(v1/n1 + v2/n2) — a signal-to-noise ratio; large \|t\| means unlikely due to chance.
4. Welch-Satterthwaite df approximates the effective degrees of freedom when variances/sizes are unequal (usually non-integer) — the noisier/smaller group contributes less reliable info, lowering usable df.
5. p-value: P(t this extreme \| H0 true), two-tailed since H1 is "different," not directional.
6. Decision: p<alpha → reject H0 (real difference); p>=alpha → fail to reject (NOT proof H0 is true, just insufficient evidence).
7. Cohen's d = (m1-m2) / pooled_SD answers "how big," not "is it real" — catches a tiny effect that's statistically significant only because n is large (~0.2 small, ~0.5 medium, ~0.8 large).
8. Report both: t-stat/p-value = is it real; Cohen's d = does it matter.

**Paired T-Test — Same Subjects, Before/After**

1. Different from an independent t-test: compute diff = after−before per pair first, then run a one-sample t-test on those diffs.
2. t = mean_diff / (sd_diff/√n), df = n-1.
3. No between-group independence assumption needed since it's the same subjects — removes between-subject variance, making it more powerful than the unpaired test for the same effect size.

**ANOVA — Comparing 3+ Groups**

1. Running pairwise t-tests inflates the false-positive rate (the multiple-testing problem) — use ANOVA (F-test) instead.
2. F = between-group variance / within-group variance — a large F means group means vary more than within-group noise explains.
3. A significant ANOVA tells you SOME group differs, not WHICH — it needs a post-hoc test (e.g. Tukey HSD) to find which pairs.

**Proportion Z-Test — Comparing Conversion Rates**

1. A different test family from the t-test: a proportion's variance is a known function of p (p(1-p)/n), not estimated separately.
2. z = (p1-p2) / sqrt(p_pooled·(1-p_pooled)·(1/n1+1/n2)), where p_pooled = the pooled conversion rate across both groups.
3. Use z, not t, since a proportion's variance doesn't need separate estimation — the standard A/B testing significance test for conversion-rate metrics.

**Confidence Intervals — What 95% Actually Means**

1. CI = range of plausible values for a population parameter, from sample data. Formula (mean): x̄ ± z(or t)·(s/√n).
2. 95% CI meaning: if we repeated sampling many times and built a CI each time, 95% of those intervals would contain the true parameter — a statement about the method's long-run reliability, not this one interval.
3. NOT "95% probability the true mean is in THIS interval" (the true mean is fixed, either in or out — no probability attaches to a fixed value). NOT "95% of the data falls in this range" (that's a different, prediction/percentile interval).
4. Intuition: like a process that's "95% reliable" — one output isn't 95% good, it's good or not; the process succeeds 95% of the time.

**Type I vs. Type II Error, and Power**

1. Type I (α): reject H0 when true — a false positive. Type II (β): fail to reject H0 when false — a false negative, missing a real effect.
2. Power = 1-β: the probability of detecting a real effect. Depends on effect size, n, and α.
3. Trade-off: lower α → fewer Type I errors, but more Type II errors at a fixed n. Only a bigger n improves both.
4. Which is worse is context-dependent: medical screening (Type II worse — missing disease costs lives); criminal trial (Type I worse — convicting the innocent); spam filter (Type I worse — flagging real email); drug approval (Type I worse — approving a harmful/useless drug); A/B test ship decision (usually Type I worse — wasting effort on a no-op change).
5. Interview framing: state which error is costlier in the given scenario, and which way that biases the design (e.g. lower α if a false positive is worse).

**Conditional Probability from Raw Counts (When Not to Use Bayes)**

1. Setup: 4,000 users on version A, 5,000 on B; 3,000 total experienced a bug, 1,500 of those on B. Want P(B \| bug).
2. Key idea: when given counts (not stated rates/probabilities), direct counting beats setting up the full Bayes formula.
3. Solve: P(B \| bug) = 1500 / 3000 = 0.5.
4. Same result via Bayes takes more steps: P(B \| bug) = P(bug\|B)·P(B) / P(bug) = (1500/5000)·(5000/9000) / (3000/9000) = 0.5 — same answer, more work.
5. Recognition signal: if a question gives overlapping raw counts (not percentages framed as "P(X\|Y)"), check whether a direct count ratio answers it before reaching for Bayes.
