#### Probability & Statistics

# Probability Foundations

## Expectation, Variance and Common Distributions

### Expectation $\mathbb{E}[X]$

The long-run average value of a random variable $X$.

* **Discrete:** $\mathbb{E}[X] = \sum x \cdot P(X = x)$
* **Continuous:** $\mathbb{E}[X] = \int_{-\infty}^{\infty} x \cdot f(x) \, dx$
* **Linearity Property:** $\mathbb{E}[aX + bY] = a\mathbb{E}[X] + b\mathbb{E}[Y]$ *(always holds, even if $X$ and $Y$ are dependent)*.

---

### Variance $\text{Var}(X)$

Measures the average squared spread of data points around the mean.

$$\text{Var}(X) = \mathbb{E}[(X - \mu)^2] = \mathbb{E}[X^2] - (\mathbb{E}[X])^2$$

* **Standard Deviation:** $\sigma = \sqrt{\text{Var}(X)}$ (in original units).
* **Properties:**
* $\text{Var}(aX + b) = a^2 \text{Var}(X)$
* If $X, Y$ are independent: $\text{Var}(X \pm Y) = \text{Var}(X) + \text{Var}(Y)$



---

### Key Distributions Summary

| Distribution | Type | PMF / PDF | Mean ($\mu$) | Variance ($\sigma^2$) | Primary Use Case |
| --- | --- | --- | --- | --- | --- |
| **Bernoulli** | Discrete | $P(X=k) = p^k (1-p)^{1-k}$ | $p$ | $p(1-p)$ | Single binary trial (Success/Failure) |
| **Binomial** | Discrete | $\binom{n}{k} p^k (1-p)^{n-k}$ | $np$ | $np(1-p)$ | Number of successes in $n$ independent trials |
| **Poisson** | Discrete | $\frac{\lambda^k e^{-\lambda}}{k!}$ | $\lambda$ | $\lambda$ | Event occurrences per fixed time/space unit |
| **Uniform** | Cont. | $\frac{1}{b - a}$ for $x \in [a, b]$ | $\frac{a+b}{2}$ | $\frac{(b-a)^2}{12}$ | Equal probability outcomes across continuous interval |
| **Normal** | Cont. | $\frac{1}{\sigma \sqrt{2\pi}} e^{-\frac{(x-\mu)^2}{2\sigma^2}}$ | $\mu$ | $\sigma^2$ | Natural measurement phenomena (Central Limit Theorem) |

---

## Bernoulli and Uniform Distributions

* **Bernoulli Distribution:** Models a single trial with two outcomes: $1$ ("Success", probability $p$) or $0$ ("Failure", probability $1-p$).
* **Continuous Uniform Distribution $U(a, b)$:** All equal-length sub-intervals within $[a, b]$ are equally likely.
* Cumulative Distribution Function (CDF): $F(x) = \frac{x - a}{b - a}$ for $x \in [a, b]$.



---

## Bayes' Theorem

Calculates the updated probability of a hypothesis $A$ given observed evidence $B$:

$$P(A \mid B) = \frac{P(B \mid A) \cdot P(A)}{P(B)}$$

Where:

* $P(A \mid B)$ = **Posterior probability** (belief after seeing evidence $B$)
* $P(B \mid A)$ = **Likelihood** (probability of evidence given hypothesis $A$)
* $P(A)$ = **Prior probability** (initial belief before evidence)
* $P(B)$ = **Marginal Likelihood / Evidence** = $\sum_i P(B \mid A_i) P(A_i)$ (normalizing constant)

---

## Covariance and Correlation

### Covariance

Measures the directional linear relationship between two variables:

$$\text{Cov}(X, Y) = \mathbb{E}[(X - \mu_X)(Y - \mu_Y)] = \mathbb{E}[XY] - \mathbb{E}[X]\mathbb{E}[Y]$$

* Scale-dependent (hard to interpret raw values).

### Pearson Correlation Coefficient ($\rho$)

Standardizes covariance into a scale-invariant metric between $-1$ and $+1$:

$$\rho_{X, Y} = \frac{\text{Cov}(X, Y)}{\sigma_X \sigma_Y}$$

* $\rho = +1$: Perfect positive linear relationship.
* $\rho = 0$: No linear relationship.
* $\rho = -1$: Perfect negative linear relationship.

---

## Descriptive Statistics and Skew

### Measures of Central Tendency & Spread

* **Mean:** Sensitivity to extreme outliers.
* **Median:** Robust central measure ($50^{\text{th}}$ percentile).
* **Mode:** Most frequent value.
* **IQR (Interquartile Range):** $Q_3 - Q_1$ (robust measure of scale).

### Skewness

Measures asymmetry of a probability distribution:

```
    Left-Skewed (Negative)             Right-Skewed (Positive)
    Mean < Median < Mode               Mode < Median < Mean
         ┌─────┐                            ┌─────┐
       ┌─┘     │                          │     └─┐
     ──┴───────┴──────                  ──┴───────┴──────

```

* **Positive / Right Skew:** Long tail extends to the right ($\text{Mean} > \text{Median}$).
* **Negative / Left Skew:** Long tail extends to the left ($\text{Mean} < \text{Median}$).

---

## Conditional Probability and Independence

* **Conditional Probability:** $P(A \mid B) = \frac{P(A \cap B)}{P(B)}$ (where $P(B) > 0$).
* **Independence:** Events $A$ and $B$ are independent if and only if:

$$P(A \cap B) = P(A) \cdot P(B) \iff P(A \mid B) = P(A)$$



---

## z-scores and Standardization

Standardization centers data to mean $\mu = 0$ and scales variance to $\sigma^2 = 1$:

$$z = \frac{x - \mu}{\sigma}$$

* **Interpretation:** A $z$-score of $+2.0$ means the observation sits $2$ standard deviations above the population mean.

---

## Classic Probability Problems

### Monty Hall Problem

* **Scenario:** 3 doors (1 car, 2 goats). You pick Door 1. Host opens Door 3 showing a goat and offers a switch to Door 2.
* **Solution:** Always **switch**.
* Initial pick win probability = $\frac{1}{3}$.
* Unchosen doors combined probability = $\frac{2}{3}$. Since the host eliminates a goat door, Door 2 inherits the full $\frac{2}{3}$ winning probability.



### Birthday Paradox

* **Scenario:** How many people needed in a room for a $\ge 50\%$ chance that at least two share a birthday?
* **Solution:** **23 people**.
* Calculate complement probability (all distinct birthdays):

$$P(\text{all distinct}) = \frac{365}{365} \times \frac{364}{365} \times \dots \times \frac{365 - n + 1}{365}$$


* For $n = 23$: $P(\text{at least 1 match}) = 1 - P(\text{all distinct}) \approx 50.7\%$.

# Estimation: Frequentist and Bayesian

## MLE vs MAP

### Maximum Likelihood Estimation (MLE)

Finds parameter values $\theta$ that maximize the probability of observing the sample data $X$:

$$\hat{\theta}_{\text{MLE}} = \arg\max_{\theta} P(X \mid \theta) = \arg\max_{\theta} \sum_{i=1}^n \log P(x_i \mid \theta)$$

### Maximum A Posteriori (MAP)

Incorporates a **prior distribution** $P(\theta)$ over parameters using Bayes' rule:

$$\hat{\theta}_{\text{MAP}} = \arg\max_{\theta} P(\theta \mid X) = \arg\max_{\theta} \left[ P(X \mid \theta) P(\theta) \right] = \arg\max_{\theta} \left[ \log P(X \mid \theta) + \log P(\theta) \right]$$

### Key Connection to Regularization

* MAP with a **Gaussian Prior** $\mathcal{N}(0, \sigma^2)$ on weights $\equiv$ **$L_2$ Regularization (Ridge)**.
* MAP with a **Laplace Prior** on weights $\equiv$ **$L_1$ Regularization (Lasso)**.

---

## Bayesian vs Frequentist

| Dimension | Frequentist Paradigm | Bayesian Paradigm |
| --- | --- | --- |
| **Parameter View** | Fixed, unknown constant parameter $\theta$. | Random variable with a probability distribution $P(\theta)$. |
| **Data View** | Random sample drawn from repeatable trials. | Observed fixed data. |
| **Intervals** | 95% Confidence Interval (in 95% of repeated samples, interval contains true fixed $\theta$). | 95% Credible Interval (there is a 95% probability $\theta$ lies within interval). |
| **Handling Prior Info** | Excluded; inferences depend strictly on observed sample $X$. | Formally integrated via Prior $P(\theta)$. |

# Sampling and Inference

## Central Limit Theorem (CLT)

Regardless of the population's underlying distribution (provided finite variance $\sigma^2$), the sampling distribution of the sample mean $\bar{X}_n$ approaches a Normal distribution as sample size $n \to \infty$ ($n \ge 30$ heuristic):

$$\bar{X}_n \sim \mathcal{N}\left(\mu, \frac{\sigma^2}{n}\right)$$

* **Standard Error (SE):** $\text{SE} = \frac{\sigma}{\sqrt{n}}$

---

## Confidence Intervals

An interval estimate constructed from sample data with a pre-specified confidence level ($1 - \alpha$):

$$\text{CI} = \bar{X} \pm \left( z_{\alpha/2} \cdot \frac{\sigma}{\sqrt{n}} \right) \quad \text{or} \quad \bar{X} \pm \left( t_{\alpha/2, n-1} \cdot \frac{s}{\sqrt{n}} \right)$$

* **95% CI ($z$-critical):** $z_{0.025} = 1.96$
* **Interpretation:** In repeated sampling, 95% of computed confidence intervals will contain the true population parameter $\mu$.

---

## Law of Large Numbers vs Central Limit Theorem

* **Law of Large Numbers (LLN):** Sample mean **converges in value** to the true population mean as sample size grows: $\bar{X}_n \to \mu$ as $n \to \infty$. (Focuses on **point convergence**).
* **Central Limit Theorem (CLT):** Describes the **shape of the distribution** of sample means around $\mu$ as $n \to \infty$. (Focuses on **distributional convergence**).

---

## Sampling and Sampling Bias

* **Probability Sampling:** Simple Random, Stratified (sample proportional to strata), Cluster Sampling.
* **Biases:**
* **Selection Bias:** Sample non-representative of target population.
* **Survivorship Bias:** Evaluating only successful instances surviving a selection process.
* **Non-Response Bias:** Systemic differences between respondents and non-respondents.



---

## Bootstrap

A non-parametric resampling technique to estimate sampling distributions without assuming underlying parametric distributions.

### Algorithm

1. Draw $B$ independent resamples (e.g., $B = 1,000$) of size $n$ **with replacement** from original dataset of size $n$.
2. Compute the metric of interest $\hat{\theta}_b^*$ for each bootstrap sample $b$.
3. Estimate standard error as standard deviation of $\{\hat{\theta}_1^*, \dots, \hat{\theta}_B^*\}$.
4. **95% Percentile CI:** Take $2.5^{\text{th}}$ and $97.5^{\text{th}}$ percentiles of bootstrap estimates.

# Hypothesis Testing

## The General Framework

1. State Null Hypothesis ($H_0$) and Alternative Hypothesis ($H_1$).
2. Select significance level $\alpha$ (typically $0.05$).
3. Compute test statistic from sample data.
4. Obtain $p$-value (probability of observing a test statistic as or more extreme under $H_0$).
5. Reject $H_0$ if $p \le \alpha$.

### Error Types

* **Type I Error ($\alpha$):** False Positive (Rejecting true $H_0$).
* **Type II Error ($\beta$):** False Negative (Failing to reject false $H_0$).
* **Statistical Power ($1 - \beta$):** Probability of correctly rejecting false $H_0$.

---

## One-Sample t-Test (Worked by Hand)

Tests if a sample mean differs significantly from a known population value $\mu_0$.

* **Data:** $X = [12, 14, 15, 18, 21]$ ($n = 5$). Test if true $\mu_0 = 12$ ($\alpha = 0.05$).

### Step 1: Compute Sample Mean ($\bar{x}$) and Standard Deviation ($s$)

$$\bar{x} = \frac{12 + 14 + 15 + 18 + 21}{5} = \frac{80}{5} = 16.0$$

$$s^2 = \frac{\sum (x_i - \bar{x})^2}{n - 1} = \frac{(-4)^2 + (-2)^2 + (-1)^2 + 2^2 + 5^2}{4} = \frac{16 + 4 + 1 + 4 + 25}{4} = \frac{50}{4} = 12.5$$

$$s = \sqrt{12.5} \approx 3.535$$

### Step 2: Compute $t$-Statistic

$$t = \frac{\bar{x} - \mu_0}{s / \sqrt{n}} = \frac{16 - 12}{3.535 / \sqrt{5}} = \frac{4}{3.535 / 2.236} = \frac{4}{1.581} \approx \mathbf{2.53}$$

### Step 3: Decision

* Degrees of freedom $df = n - 1 = 4$.
* Critical $t$-value ($df=4, \alpha=0.05$, two-tailed) = $2.776$.
* Since $\vert{}t_{\text{calc}}\vert{} = 2.53 < 2.776$, **fail to reject $H_0$**.

---

## Two-Sample t-Test (Comparing Means)

Tests if two independent population means differ ($\mu_1 - \mu_2 = 0$).

### Formula (Pooled Variance equal-variance assumption)

$$t = \frac{\bar{x}_1 - \bar{x}_2}{s_p \sqrt{\frac{1}{n_1} + \frac{1}{n_2}}}$$

Where pooled variance $s_p^2 = \frac{(n_1 - 1)s_1^2 + (n_2 - 1)s_2^2}{n_1 + n_2 - 2}$ ($df = n_1 + n_2 - 2$).

---

## Chi-Square Test ($\chi^2$) (Worked by Hand)

Tests goodness of fit or independence between two categorical variables.

* **Formula:** $\chi^2 = \sum \frac{(O - E)^2}{E}$

### Worked Example (Goodness of Fit)

Equal distribution across 3 categories ($H_0: p_1 = p_2 = p_3 = 1/3$). Total $N = 150$.

* Observed Counts ($O$): $A = 60, B = 45, C = 45$.
* Expected Counts ($E$): $150 / 3 = 50$ for each category.

### Step-by-Step Calculation

$$\chi^2 = \frac{(60 - 50)^2}{50} + \frac{(45 - 50)^2}{50} + \frac{(45 - 50)^2}{50}$$

$$\chi^2 = \frac{100}{50} + \frac{25}{50} + \frac{25}{50} = 2.0 + 0.5 + 0.5 = \mathbf{3.0}$$

* $df = k - 1 = 3 - 1 = 2$.
* Critical $\chi^2_{df=2, \alpha=0.05} = 5.991$.
* Since $3.0 < 5.991$, **fail to reject $H_0$**.

---

## ANOVA (Worked by Hand - Comparing 3+ Groups)

Tests if means across $k \ge 3$ groups are equal without inflating Type I error rate.

* **Data:**
* Group 1: $[2, 4, 6]$ ($\bar{x}_1 = 4, n_1 = 3$)
* Group 2: $[4, 6, 8]$ ($\bar{x}_2 = 6, n_2 = 3$)
* Group 3: $[6, 8, 10]$ ($\bar{x}_3 = 8, n_3 = 3$)


* **Grand Mean ($\bar{X}$):** $\frac{4 + 6 + 8}{3} = 6.0$, Total $N = 9$.

### Step 1: Sum of Squares Between ($SS_{\text{Between}}$)

$$SS_{\text{Between}} = \sum n_j (\bar{x}_j - \bar{X})^2 = 3(4-6)^2 + 3(6-6)^2 + 3(8-6)^2 = 3(4) + 0 + 3(4) = \mathbf{24.0}$$

$$df_{\text{Between}} = k - 1 = 3 - 1 = 2 \implies MS_{\text{Between}} = \frac{24}{2} = \mathbf{12.0}$$

### Step 2: Sum of Squares Within ($SS_{\text{Within}}$)

Calculate variance within each group around its own group mean:

* Group 1: $(2-4)^2 + (4-4)^2 + (6-4)^2 = 4 + 0 + 4 = 8$
* Group 2: $(4-6)^2 + (6-6)^2 + (8-6)^2 = 4 + 0 + 4 = 8$
* Group 3: $(6-8)^2 + (8-8)^2 + (10-8)^2 = 4 + 0 + 4 = 8$

$$SS_{\text{Within}} = 8 + 8 + 8 = \mathbf{24.0}$$

$$df_{\text{Within}} = N - k = 9 - 3 = 6 \implies MS_{\text{Within}} = \frac{24}{6} = \mathbf{4.0}$$

### Step 3: Compute $F$-Statistic

$$F = \frac{MS_{\text{Between}}}{MS_{\text{Within}}} = \frac{12.0}{4.0} = \mathbf{3.0}$$

* Critical $F(2, 6, \alpha=0.05) = 5.14$.
* Since $F_{\text{calc}} = 3.0 < 5.14$, **fail to reject $H_0$**.

---

## Normality Tests & Kolmogorov-Smirnov (KS) Test

### Common Normality Tests

1. **Shapiro-Wilk Test:** High statistical power for small sample sizes ($n < 50$).
2. **D'Agostino-Pearson Test:** Evaluates skewness and kurtosis.
3. **Kolmogorov-Smirnov (KS) Test:** Non-parametric test comparing sample empirical CDF against a reference theoretical CDF (or comparing two empirical CDFs):

$$D = \max_x \vert{}F_{\text{obs}}(x) - F_{\text{exp}}(x)\vert{}$$



---

## One-Tailed vs Two-Tailed Tests

* **Two-Tailed Test:** $H_1: \mu \neq \mu_0$. Rejects $H_0$ if test statistic is in either extreme tail ($\alpha/2$ area per tail).
* **One-Tailed Test:** $H_1: \mu > \mu_0$ or $H_1: \mu < \mu_0$. Used only when direction is hypothesized *a priori*. Rejects $H_0$ in a single direction (full $\alpha$ area in one tail).

---

## Non-Parametric Tests

Used when distributional assumptions (normality) are violated.

| Parametric Test | Non-Parametric Equivalent | Application |
| --- | --- | --- |
| **One-sample t-test** | Wilcoxon Signed-Rank Test | Median vs Hypothesized Value |
| **Two-sample t-test** | Mann-Whitney U Test | Comparing 2 Independent Medians |
| **Paired t-test** | Wilcoxon Signed-Rank Test | Comparing 2 Paired/Matched Samples |
| **One-Way ANOVA** | Kruskal-Wallis Test | Comparing $k \ge 3$ Independent Groups |

---

## Multiple Comparisons: FWER and FDR

Testing multiple hypotheses simultaneously increases overall Type I error probability (Family-Wise Error Rate): $P(\text{at least 1 False Positive}) = 1 - (1 - \alpha)^m$.

### 1. Family-Wise Error Rate (FWER) Controls

Controls probability of making **at least one** Type I error across $m$ tests.

* **Bonferroni Correction:** Adjusts threshold $\alpha_{\text{new}} = \frac{\alpha}{m}$.
* *Trait:* Highly conservative; increases Type II error rate.



### 2. False Discovery Rate (FDR) Controls

Controls expected proportion of false rejections among all rejections ($\frac{FP}{FP + TP}$).

* **Benjamini-Hochberg (BH) Procedure:**
1. Rank $p$-values in ascending order: $p_{(1)} \le p_{(2)} \le \dots \le p_{(m)}$.
2. Find largest index $k$ where $p_{(k)} \le \frac{k}{m} q^*$.
3. Reject $H_0$ for all tests $i = 1, \dots, k$.


* *Trait:* Higher statistical power than Bonferroni; ideal for high-throughput experiments (A/B testing platforms, genomics).
