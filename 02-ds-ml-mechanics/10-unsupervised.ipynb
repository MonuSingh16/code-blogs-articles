#### Clustering Algorithms

**1. K-Means & K-Means++** - $K$-Means partitions a dataset of $N$ points into $K$ distinct, non-overlapping clusters. Each cluster is defined by its center point, called a centroid ($\mu_k$).

1. The objective is to minimize the Within-Cluster Sum of Squares (WCSS), also known as Inertia:$$J = \sum_{k=1}^K \sum_{i \in C_k} \Vert{}\mathbf{x}_i - \boldsymbol{\mu}_k\Vert{}^2$$ . where:$C_k$ is the set of data points assigned to cluster $k$.$\boldsymbol{\mu}_k = \frac{1}{\vert{}C_k\vert{}} \sum_{i \in C_k} \mathbf{x}_i$ is the mean (centroid) of cluster $k$.

2. Steps: 
    1. Initialization: Randomly pick $K$ data points as initial centroids $\boldsymbol{\mu}_1, \dots, \boldsymbol{\mu}_K$.
    2. Assignment Step: Assign each data point $\mathbf{x}_i$ to its closest centroid using Euclidean distance:$$c_i = \arg\min_k \Vert{}\mathbf{x}_i - \boldsymbol{\mu}_k\Vert{}^2$$
    3. Update Step: Recompute each centroid as the mean of all data points assigned to its cluster:$$\boldsymbol{\mu}_k = \frac{1}{\vert{}C_k\vert{}} \sum_{i \in C_k} \mathbf{x}_i$$
    4. Convergence: Repeat Steps 2 and 3 until centroids stop moving, point assignments no longer change, or a maximum iteration limit is reached.

3. Standard $K$-Means selects initial centroids purely at random. If two initial centroids happen to be picked close to each other, the algorithm can easily get trapped in a poor local minimum, leading to sub-optimal clustering. --> The $K$-Means++ : Initializes centroids smartly by spreading them out across the feature space before running Lloyd's algorithm.
    1. Choose the first centroid $\boldsymbol{\mu}_1$ uniformly at random from the data points.
    2. For each data point $\mathbf{x}_i$, compute its distance $D(\mathbf{x}_i)$ to the nearest already chosen centroid.
    3. Choose the next centroid $\boldsymbol{\mu}_k$ randomly from the data points with probability proportional to $D(\mathbf{x}_i)^2$:$$P(\mathbf{x}_i) = \frac{D(\mathbf{x}_i)^2}{\sum_{j=1}^N D(\mathbf{x}_j)^2}$$
    4. Repeat Steps 2 and 3 until all $K$ centroids are chosen.

4. The Elbow Method - Plot Inertia (WCSS) on the Y-axis against different values of $K$ (e.g., $K \in [1, 10]$) on the X-axis. As $K$ increases, Inertia monotonically decreases toward 0 (at $K=N$, Inertia is 0). Look for the "elbow" point—the point where the rate of decrease drastically slows down. 

5. Silhouette Score Measures how well-separated clusters are relative to their internal density. For a sample $i$:$$s_i = \frac{b_i - a_i}{\max(a_i, b_i)}$$ $a_i$: Mean distance from sample $i$ to all other points in the same cluster (intra-cluster distance).$b_i$: Mean distance from sample $i$ to all points in the nearest neighboring cluster (inter-cluster distance). $+1$: Sample is well-clustered and far from neighboring clusters. $0$: Sample sits right on the decision boundary between two clusters.$-1$: Sample has been misclustered.

6. Assumes Spherical Clusters: $K$-Means uses Euclidean distance, assuming clusters are isotropic (spherical) and of roughly equal variance/size.
