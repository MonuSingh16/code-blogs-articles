**1. CV Techniques**

1. Standard $K$-Fold Cross-ValidationLogic: Splits dataset into $K$ equal parts. Trains on $K-1$ folds and validates on the remaining fold, repeating $K$ times. Unsuitable for imbalanced datasets or grouped/sequential data.

2. Stratified $K$-FoldLogic: Ensures each fold maintains the exact same percentage of target class labels as the complete dataset. When to use: Imbalanced classification problems (e.g., 99:1 target ratio).

3. Group $K$-FoldLogic: Prevents samples from the same logical group/entity (e.g., multiple patient scans from the same individual, or multiple transactions from the same user) from appearing in both train and validation folds simultaneously.When to use: Multi-observation data per entity (prevents group identity leakage).

4. Time-Series Split (Forward Chaining / Rolling Window)Logic: Enforces temporal causality by training strictly on past observations to predict future observations.$$\text{Fold 1: } [T_1 \dots T_2] \rightarrow \text{Val: } [T_3]$$ $$\text{Fold 2: } [T_1 \dots T_3] \rightarrow \text{Val: } [T_4]$$ When to use: Financial time-series, demand forecasting, stock prices. Using standard $K$-Fold on time-series data causes look-ahead leakage (training on future data to predict the past).

> Preprocessing Leakage: Avoid Scaling, normalization, or mean imputation performed on the entire dataset before CV splitting. 

> Fix: Fit transformations only on training folds inside a Scikit-Learn Pipeline

**2. Hyperparameter Tuning**

1. Grid Search (GridSearchCV) Mechanism: Exhaustively evaluates every parameter combination across a manually specified grid . Extremely slow; suffers from the curse of dimensionality as parameter options increase linearly ($O(\prod n_i)$ trials).

2. Random Search (RandomizedSearchCV) Mechanism: Samples fixed $N$ parameter combinations randomly from specified statistical distributions. Far more efficient than Grid Search; focuses compute resources on sensitive hyperparameters. Cons: May miss optimal combinations in narrow hyperparameter regions.

3. Bayesian Optimization (Optuna, Hyperopt) Mechanism: Builds a probabilistic surrogate model (e.g., Tree-structured Parzen Estimator / TPE) of the objective function to predict which hyperparameter combinations are most likely to improve performance based on previous trial outcomes. Pros: Highly efficient; converges to optimal parameters in fewer iterations. Cons: Sequential optimization limits parallel compute compared to pure random search.
